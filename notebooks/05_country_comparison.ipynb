# Five-Year Budget Impact Comparison

This visualisation compares the projected financial impact of semaglutide adoption under UK and US healthcare system assumptions.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Country-level net budget impact
comparison = pd.DataFrame({
    "Country": ["United Kingdom", "United States", "Italy"],
    "Currency": ["GBP", "USD", "EUR"],
    "Year 1": [
        394_912_800,
        67_469_563_908.79,
        890_793_900
    ],
    "Year 5": [
        1_974_564_000,
        337_347_819_543.94,
        4_453_969_500
    ]
})

# Convert to millions for readability
comparison["Year 1 (million)"] = comparison["Year 1"] / 1_000_000
comparison["Year 5 (million)"] = comparison["Year 5"] / 1_000_000

display(
    comparison[
        ["Country", "Currency", "Year 1 (million)", "Year 5 (million)"]
    ]
)

# Create comparative figure
fig, ax = plt.subplots(figsize=(10, 6))

x = range(len(comparison))
width = 0.35

ax.bar(
    [i - width/2 for i in x],
    comparison["Year 1 (million)"],
    width=width,
    label="Year 1"
)

ax.bar(
    [i + width/2 for i in x],
    comparison["Year 5 (million)"],
    width=width,
    label="Year 5"
)

ax.set_xticks(list(x))
ax.set_xticklabels(
    [
        "United Kingdom\n(GBP)",
        "United States\n(USD)",
        "Italy\n(EUR)"
    ]
)

ax.set_ylabel("Net budget impact (million local currency units)")
ax.set_title("Projected Semaglutide Budget Impact: UK, USA and Italy")
ax.legend()

plt.tight_layout()

# Create comparative reports folder if needed
comparative_dir = Path("../reports/comparative")
comparative_dir.mkdir(parents=True, exist_ok=True)

# Save figure
figure_path = comparative_dir / "country_budget_impact_comparison.png"
plt.savefig(figure_path, dpi=300, bbox_inches="tight")

plt.show()

print(f"Figure saved to: {figure_path}")
```

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# STEP 68 — REGENERATE UK BUDGET IMPACT FIGURE
# ---------------------------------------------------------

uk_figures_dir = Path("../reports/uk/figures")
uk_figures_dir.mkdir(parents=True, exist_ok=True)

# UK model values
uk_eligible_population = 3_290_941
uk_annual_drug_cost = 2_500

# Uptake assumptions
uptake = {
    "Year 1": 0.05,
    "Year 2": 0.10,
    "Year 3": 0.15,
    "Year 4": 0.20,
    "Year 5": 0.25
}

# Calculate treated population and budget impact
uk_results = []

for year, rate in uptake.items():
    treated = round(uk_eligible_population * rate)
    budget_impact = treated * uk_annual_drug_cost

    uk_results.append({
        "Year": year,
        "Uptake": rate,
        "Treated Population": treated,
        "Budget Impact": budget_impact
    })

uk_df = pd.DataFrame(uk_results)

# ---------------------------------------------------------
# Create figure
# ---------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.plot(
    uk_df["Year"],
    uk_df["Budget Impact"] / 1_000_000,
    marker="o",
    linewidth=2.5
)

plt.title(
    "UK Semaglutide Budget Impact by Year",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Year", fontsize=12)
plt.ylabel("Budget Impact (£ million)", fontsize=12)

plt.grid(True, alpha=0.3)

plt.tight_layout()

# Save as PNG
output_path = uk_figures_dir / "uk_budget_impact.png"

plt.savefig(
    output_path,
    dpi=300,
    bbox_inches="tight",
    format="png"
)

plt.close()

print("UK budget impact figure regenerated.")
print(f"Saved to: {output_path}")
print(f"File exists: {output_path.exists()}")
print(f"File size: {output_path.stat().st_size / 1024:.1f} KB")
```

```python
from pathlib import Path
import pandas as pd

# ---------------------------------------------------------
# STEP 74A — CREATE UK BUDGET IMPACT SUMMARY
# ---------------------------------------------------------

uk_tables = Path("../reports/uk/tables")
uk_tables.mkdir(parents=True, exist_ok=True)

# UK model assumptions
eligible_population = 3_290_941
annual_drug_cost = 2_500

uptake = {
    "Year 1": 0.05,
    "Year 2": 0.10,
    "Year 3": 0.15,
    "Year 4": 0.20,
    "Year 5": 0.25
}

results = []

for year, uptake_rate in uptake.items():

    treated_population = round(
        eligible_population * uptake_rate
    )

    budget_impact = (
        treated_population * annual_drug_cost
    )

    results.append({
        "Year": year,
        "Uptake": uptake_rate,
        "Eligible Population": eligible_population,
        "Treated Population": treated_population,
        "Annual Drug Cost": annual_drug_cost,
        "Budget Impact": budget_impact
    })

uk_budget_df = pd.DataFrame(results)

output_file = uk_tables / "uk_budget_impact_summary.csv"

uk_budget_df.to_csv(
    output_file,
    index=False
)

print("UK budget impact summary created.")
print(f"Saved to: {output_file}")
print()
print(uk_budget_df)
```
