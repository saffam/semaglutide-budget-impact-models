# Country Comparison Dataset

This section combines the completed UK and USA budget impact models into a single comparative dataset.

```python
comparison = pd.DataFrame({
    "Metric": [
        "Adult population",
        "Obesity prevalence",
        "Eligible population",
        "Annual treatment cost",
        "Year 1 budget impact",
        "Year 5 budget impact"
    ],
    "United Kingdom": [
        "55.0 million",
        "29.9%",
        "3.29 million",
        "£2,500",
        "£394.9 million",
        "£1.97 billion"
    ],
    "United States": [
        "269.8 million",
        "33.1%",
        "89.2 million",
        "$17,537",
        "$67.5 billion",
        "$337.3 billion"
    ]
})

comparison
```

```python
comparison_path = (
    project_root /
    "reports" /
    "comparison" /
    "uk_vs_usa_summary.csv"
)

comparison_path.parent.mkdir(
    parents=True,
    exist_ok=True
)

comparison.to_csv(
    comparison_path,
    index=False
)

print("Saved:", comparison_path)
```

```python
import matplotlib.pyplot as plt

countries = ["United Kingdom", "United States"]

eligible_population = [
    3.29,
    89.2
]

plt.figure(figsize=(8,5))

plt.bar(
    countries,
    eligible_population
)

plt.ylabel("Eligible Population (millions)")
plt.title(
    "Estimated Eligible Population for Semaglutide Treatment"
)

plt.xticks(rotation=20)

plt.tight_layout()

plt.savefig(
    project_root /
    "reports" /
    "comparison" /
    "eligible_population_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

# Five-Year Budget Impact Comparison

This visualisation compares the projected financial impact of semaglutide adoption under UK and US healthcare system assumptions.

```python
import matplotlib.pyplot as plt

countries = [
    "United Kingdom",
    "United States"
]

year_1_budget = [
    394.9,
    67469.56
]

year_5_budget = [
    1974.56,
    337347.82
]

plt.figure(figsize=(8,5))

x = range(len(countries))

plt.bar(
    x,
    year_1_budget,
    width=0.35,
    label="Year 1"
)

plt.bar(
    [i + 0.35 for i in x],
    year_5_budget,
    width=0.35,
    label="Year 5"
)

plt.xticks(
    [i + 0.175 for i in x],
    countries,
    rotation=20
)

plt.ylabel("Net Budget Impact (million currency units)")

plt.title(
    "Projected Semaglutide Budget Impact: UK vs USA"
)

plt.legend()

plt.tight_layout()

plt.savefig(
    project_root /
    "reports" /
    "comparison" /
    "budget_impact_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
