```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
uk = pd.read_csv(
"../reports/uk/tables/uk_model_summary.csv"
)

uk
```

```python
drug_prices = [
2000,
2500,
3000,
3500
]

results=[]

for price in drug_prices:

    impact = (
        3290941 *
        0.05 *
        price
    )

    results.append(
        {
        "Drug Cost":price,
        "Year 1 Budget Impact":impact
        }
    )


sensitivity = pd.DataFrame(results)

sensitivity
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    sensitivity["Drug Cost"],
    sensitivity["Year 1 Budget Impact"],
    marker="o"
)

plt.xlabel("Annual Drug Cost (£)")
plt.ylabel("Budget Impact (£)")

plt.title(
    "UK Semaglutide Budget Impact Sensitivity Analysis"
)

plt.grid()

# Save figure
plt.savefig(
    "../reports/uk/figures/uk_price_sensitivity.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
