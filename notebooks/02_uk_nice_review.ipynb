## Key Information

### NICE Recommendations

Semaglutide is recommended only when:

- used alongside a reduced-calorie diet
- increased physical activity is provided
- treatment is delivered within a specialist weight-management service
- treatment lasts for a maximum of 2 years
- patient has at least one weight-related comorbidity
- BMI ≥35 kg/m²
- or BMI 30–34.9 kg/m² if referral criteria are met

Lower BMI thresholds apply to some ethnic groups.

Treatment should be considered for stopping if weight loss is less than 5% after 6 months.
