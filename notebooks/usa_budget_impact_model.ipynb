```python
state_population = usa_population[
    (usa_population["SUMLEV"] == 40) &
    (usa_population["STATE"] != 72)  # Exclude Puerto Rico
][["NAME", "POPESTIMATE2025"]]

state_population.head()
```

```python
usa_obesity = usa_obesity.rename(columns={"State": "NAME"})

usa_weighted = state_population.merge(
    usa_obesity,
    on="NAME",
    how="inner"
)

usa_weighted.head()
```

```python
print("States matched:", len(usa_weighted))
print(usa_weighted.tail())
```

```python
# Make sure the calculation columns are numeric
usa_weighted["POPESTIMATE2025"] = pd.to_numeric(
    usa_weighted["POPESTIMATE2025"],
    errors="coerce"
)

usa_weighted["Prevalence"] = pd.to_numeric(
    usa_weighted["Prevalence"],
    errors="coerce"
)

# Remove rows where either value is missing
usa_weighted = usa_weighted.dropna(
    subset=["POPESTIMATE2025", "Prevalence"]
)

# Calculate population-weighted obesity prevalence
usa_weighted["Weighted"] = (
    usa_weighted["POPESTIMATE2025"]
    * usa_weighted["Prevalence"]
)

usa_obesity_prevalence = (
    usa_weighted["Weighted"].sum()
    / usa_weighted["POPESTIMATE2025"].sum()
)

print(f"Weighted obesity prevalence: {usa_obesity_prevalence:.2f}%")
```

```python
print("States used:", len(usa_weighted))
print(f"Weighted obesity prevalence: {usa_obesity_prevalence:.2f}%")
```

```python
matched_names = set(usa_weighted["NAME"])

missing_states = state_population[
    ~state_population["NAME"].isin(matched_names)
]

missing_states
```
