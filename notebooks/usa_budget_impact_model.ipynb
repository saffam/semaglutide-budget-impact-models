```python
obesity_file = (
    project_root
    / "data"
    / "usa"
    / "epidemiology"
    / "cdc_obesity_2024.csv"
)

usa_obesity = pd.read_csv(obesity_file)

usa_obesity.head()
```

```python
usa_obesity.columns
```

```python
usa_obesity.head(10)
```

```python
state_population = usa_population[
    (usa_population["SUMLEV"] == 40) &
    (usa_population["STATE"] != 72)  # Exclude Puerto Rico
][["NAME", "POPESTIMATE2025"]]

state_population.head()
```

```python
usa_obesity = usa_obesity.rename(columns={"State": "NAME"})

usa_weighted = state_population.merge(
    usa_obesity,
    on="NAME",
    how="inner"
)

usa_weighted.head()
```

```python
print("States matched:", len(usa_weighted))
print(usa_weighted.tail())
```

```python
# Make sure the calculation columns are numeric
usa_weighted["POPESTIMATE2025"] = pd.to_numeric(
    usa_weighted["POPESTIMATE2025"],
    errors="coerce"
)

usa_weighted["Prevalence"] = pd.to_numeric(
    usa_weighted["Prevalence"],
    errors="coerce"
)

# Remove rows where either value is missing
usa_weighted = usa_weighted.dropna(
    subset=["POPESTIMATE2025", "Prevalence"]
)

# Calculate population-weighted obesity prevalence
usa_weighted["Weighted"] = (
    usa_weighted["POPESTIMATE2025"]
    * usa_weighted["Prevalence"]
)

usa_obesity_prevalence = (
    usa_weighted["Weighted"].sum()
    / usa_weighted["POPESTIMATE2025"].sum()
)

print(f"Weighted obesity prevalence: {usa_obesity_prevalence:.2f}%")
```

```python
print("States used:", len(usa_weighted))
print(f"Weighted obesity prevalence: {usa_obesity_prevalence:.2f}%")
```

```python
matched_names = set(usa_weighted["NAME"])

missing_states = state_population[
    ~state_population["NAME"].isin(matched_names)
]

missing_states
```

```python
usa_obesity[
    ~usa_obesity["NAME"].isin(matched_names)
]
```

```python
usa_adults_with_obesity = (
    usa_adults * (usa_obesity_prevalence / 100)
)

print(f"Adults with obesity: {usa_adults_with_obesity:,.0f}")
```

```python
usa_epidemiology_summary = pd.DataFrame({
    "Measure": [
        "Adult population (18+)",
        "Population-weighted obesity prevalence",
        "Adults with obesity"
    ],
    "Value": [
        usa_adults,
        usa_obesity_prevalence,
        usa_adults_with_obesity
    ]
})

usa_epidemiology_summary
```

```python
usa_epidemiology_summary
```

```python
# USA Wegovy drug cost assumptions

usa_wegovy_monthly_cost = 1349.02
usa_months_per_year = 13

usa_wegovy_annual_cost = (
    usa_wegovy_monthly_cost * usa_months_per_year
)

print(f"Monthly Wegovy cost: ${usa_wegovy_monthly_cost:,.2f}")
print(f"Annual Wegovy cost: ${usa_wegovy_annual_cost:,.2f}")
```

```python
usa_cost_summary = pd.DataFrame({
    "Parameter": [
        "Wegovy monthly package cost",
        "Treatment cycles per year",
        "Annual drug cost per patient"
    ],
    "Value": [
        usa_wegovy_monthly_cost,
        usa_months_per_year,
        usa_wegovy_annual_cost
    ]
})

usa_cost_summary
```

```python
usa_cost_summary.to_csv(
    project_root
    / "reports"
    / "usa"
    / "tables"
    / "usa_drug_cost_summary.csv",
    index=False
)
```

```python
# USA treatment uptake assumptions
# Same scenario assumptions used in the UK model

usa_uptake_year1 = 0.05
usa_uptake_year5 = 0.25

usa_uptake_summary = pd.DataFrame({
    "Year": [1, 5],
    "Treatment_uptake": [
        usa_uptake_year1,
        usa_uptake_year5
    ]
})

usa_uptake_summary
```

```python
usa_uptake_summary.to_csv(
    project_root
    / "reports"
    / "usa"
    / "tables"
    / "usa_treatment_uptake_summary.csv",
    index=False
)
```

```python
usa_treated_year1 = (
    usa_adults_with_obesity
    * usa_uptake_year1
)

usa_treated_year5 = (
    usa_adults_with_obesity
    * usa_uptake_year5
)

print(f"Year 1 treated population: {usa_treated_year1:,.0f}")
print(f"Year 5 treated population: {usa_treated_year5:,.0f}")
```

```python
usa_budget_year1 = (
    usa_treated_year1
    * usa_wegovy_annual_cost
)

usa_budget_year5 = (
    usa_treated_year5
    * usa_wegovy_annual_cost
)

print(f"Year 1 gross drug cost: ${usa_budget_year1:,.2f}")
print(f"Year 5 gross drug cost: ${usa_budget_year5:,.2f}")
```

```python
usa_model_summary = pd.DataFrame({
    "Measure": [
        "Adult population (18+)",
        "Obesity prevalence",
        "Adults with obesity",
        "Annual drug cost per patient",
        "Treatment uptake Year 1",
        "Treatment uptake Year 5",
        "Treated population Year 1",
        "Treated population Year 5",
        "Gross drug cost Year 1",
        "Gross drug cost Year 5"
    ],
    "Value": [
        usa_adults,
        usa_obesity_prevalence,
        usa_adults_with_obesity,
        usa_wegovy_annual_cost,
        usa_uptake_year1,
        usa_uptake_year5,
        usa_treated_year1,
        usa_treated_year5,
        usa_budget_year1,
        usa_budget_year5
    ]
})

usa_model_summary
```

```python
usa_model_summary.to_csv(
    project_root
    / "reports"
    / "usa"
    / "tables"
    / "usa_model_summary.csv",
    index=False
)
```
