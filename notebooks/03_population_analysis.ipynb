```python
# Semaglutide cost assumptions

annual_cost_per_patient = 2500   # £ per patient per year (placeholder assumption)

treatment_uptake = 0.05  # 5% of eligible patients treated

eligible_population = uk_semaglutide_population.loc[0, "population"]

treated_population = int(
    eligible_population * treatment_uptake
)

total_drug_cost = (
    treated_population * annual_cost_per_patient
)

print("Eligible population:", eligible_population)
print("Treated population:", treated_population)
print("Annual drug cost (£):", total_drug_cost)
```

```python
uk_budget_impact = pd.DataFrame({
    "country": ["UK"],
    "eligible_population": [eligible_population],
    "treated_population": [treated_population],
    "annual_cost_per_patient": [annual_cost_per_patient],
    "total_drug_cost": [total_drug_cost]
})

uk_budget_impact
```

```python
uk_budget_impact.to_csv(
    "data/uk/budget_impact/uk_semaglutide_budget_impact.csv",
    index=False
)
```

```python
years = [2026, 2027, 2028, 2029, 2030]

uptake_rates = [0.05, 0.10, 0.15, 0.20, 0.25]

budget_results = []

for year, uptake in zip(years, uptake_rates):

    patients = int(
        eligible_population * uptake
    )

    cost = (
        patients * annual_cost_per_patient
    )

    budget_results.append(
        [
            year,
            uptake,
            patients,
            cost
        ]
    )


uk_5year_budget = pd.DataFrame(
    budget_results,
    columns=[
        "year",
        "uptake_rate",
        "treated_patients",
        "annual_cost"
    ]
)

uk_5year_budget
```

```python
obesity_healthcare_cost_per_patient = 500

healthcare_cost_reduction = 0.20

saving_per_patient = (
    obesity_healthcare_cost_per_patient 
    * healthcare_cost_reduction
)

saving_per_patient
```

```python
uk_5year_budget["healthcare_savings"] = (
    uk_5year_budget["treated_patients"]
    * saving_per_patient
)

uk_5year_budget
```

```python
uk_5year_budget["net_budget_impact"] = (
    uk_5year_budget["annual_cost"]
    -
    uk_5year_budget["healthcare_savings"]
)

uk_5year_budget
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.plot(
    uk_5year_budget["year"],
    uk_5year_budget["net_budget_impact"],
    marker="o"
)

plt.xlabel("Year")
plt.ylabel("Net budget impact (£)")
plt.title("UK Semaglutide Budget Impact (2026-2030)")

plt.xticks(
    uk_5year_budget["year"]
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    uk_5year_budget["year"],
    uk_5year_budget["annual_cost"],
    marker="o",
    label="Semaglutide cost"
)

plt.plot(
    uk_5year_budget["year"],
    uk_5year_budget["healthcare_savings"],
    marker="o",
    label="Healthcare savings"
)

plt.xlabel("Year")
plt.ylabel("£")
plt.title("Drug Costs vs Healthcare Savings")

plt.legend()

plt.show()
```
