# Semaglutide Budget Impact Models

## Project Goal

To build transparent budget impact models for semaglutide (Wegovy) using only official public data.

Countries:

- United Kingdom
- United States
- Italy
- Europe

---

## Main Research Question

What is the expected budget impact of introducing semaglutide for eligible adults with obesity?

---

## Rules

- Only public data
- No invented numbers
- Every result must be traceable to an official source
- Python will be used for all analyses

## UK Analysis Scope

### Decision-maker

NHS England

### Intervention

Semaglutide, sold under the brand name Wegovy.

### Indication

Weight management in adults with overweight or obesity.

This project does not analyse semaglutide for type 2 diabetes.

### Model perspective

NHS payer perspective.

### Geographic scope

England.

The analysis will initially use England rather than the entire United Kingdom because NHS commissioning, population statistics and obesity datasets are commonly reported separately for England.

### Time horizon

To be determined from official NICE and NHS evidence.

### Comparator

To be determined after reviewing the NICE appraisal and current specialist weight-management pathway.

## NICE Eligibility Criteria

According to NICE Technology Appraisal TA875, semaglutide is recommended for adults only when:

- it is provided alongside a reduced-calorie diet and increased physical activity;
- it is used within a specialist weight-management service;
- treatment lasts for a maximum of 2 years;
- the person has at least one weight-related comorbidity; and
- the relevant BMI criteria are met.

The exact BMI thresholds and population restrictions will be copied from the official guidance before modelling the eligible population.

### Marketing-authorisation population

The broader marketing authorisation covers adults with:

- BMI of at least 30 kg/m²; or
- BMI from 27 kg/m² to below 30 kg/m² with at least one weight-related comorbidity.

The NHS-recommended population may be narrower than the full marketing-authorisation population. The budget model will therefore use NICE eligibility rather than assuming that everyone covered by the licence receives NHS treatment.
